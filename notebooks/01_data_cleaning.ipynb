# Oil Sales Data Cleaning

This notebook focuses on cleaning and preparing the oil sales dataset for analysis. We'll perform the following steps:
1. Load the dataset
2. Check data quality
3. Handle missing values
4. Fix data types
5. Remove duplicates
6. Save the cleaned dataset

```python
# Import required libraries
import pandas as pd
import numpy as np

# Read the dataset
df = pd.read_csv('../data/oil_sales_assignment_dataset.csv')
```

```python
# Check basic information about the dataset
print("Dataset Info:")
print("-" * 50)
df.info()

print("\nMissing Values:")
print("-" * 50)
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

## Data Cleaning Steps

1. Convert size column from string (e.g., '0.6L') to float
2. Handle any data type conversions
3. Save the cleaned dataset

```python
# Convert size column from string (e.g., '0.6L') to float
df['size'] = df['size'].str.replace('L', '').astype(float)

# Verify the conversion
print("Size column unique values after conversion:")
print(df['size'].unique())
print("\nSize column data type:", df['size'].dtype)
```

```python
# Save the cleaned dataset
df.to_csv('../data/cleaned_data.csv', index=False)
print("Cleaned data saved to '../data/cleaned_data.csv'")
```
